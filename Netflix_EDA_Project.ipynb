```python
import pandas as pd

df = pd.read_csv('netflix_titles.csv')

df.head()
```

```python
print("Shape of dataset:", df.shape)

print("\nColumns:")
print(df.columns)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
df['director'] = df['director'].fillna('Unknown')

df['cast'] = df['cast'].fillna('Unknown')

df['country'] = df['country'].fillna('Unknown')

df['date_added'] = df['date_added'].fillna('Not Available')

df['rating'] = df['rating'].fillna('Not Rated')

df['duration'] = df['duration'].fillna('Unknown')
```

```python
print(df.isnull().sum())
```

```python
df['type'].value_counts()
```

```python
import matplotlib.pyplot as plt

df['type'].value_counts().plot(kind='bar')

plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Type")
plt.ylabel("Count")

plt.show()
```

```python
df['release_year'].value_counts().head(10)
```

```python
top_years = df['release_year'].value_counts().head(10)

top_years.sort_index().plot(kind='line')

plt.title("Top Release Years on Netflix")
plt.xlabel("Year")
plt.ylabel("Number of Releases")

plt.show()
```

```python
df['country'].value_counts().head(10)
```

```python
top_countries = df['country'].value_counts().head(5)

top_countries.plot(kind='pie', autopct='%1.1f%%')

plt.title("Top 5 Countries on Netflix")
plt.ylabel("")

plt.show()
```

```python
df['rating'].value_counts().head(10)
```

```python
df['release_year'].plot(kind='hist', bins=20)

plt.title("Distribution of Release Years")
plt.xlabel("Release Year")

plt.show()
```

```python
df['listed_in'].value_counts().head(10)
```

```python
top_genres = df['listed_in'].value_counts().head(10)

top_genres.plot(kind='barh')

plt.title("Top 10 Netflix Genres")
plt.xlabel("Count")
plt.ylabel("Genre")

plt.show()
```

```python
import seaborn as sns

numeric_df = df.select_dtypes(include=['int64', 'float64'])

sns.heatmap(numeric_df.corr(), annot=True)

plt.title("Correlation Heatmap")

plt.show()
```
